## Applying Feature Engineering on our dataset

```python
import pandas as pd
import numpy as np
from configs.settings import project_dir
```

```python
p_dir = project_dir()
clean_data_dir = p_dir.CLEANED_DIR

category_name = clean_data_dir/'category_name.csv'
customers = clean_data_dir/'clean_customers.csv'
location = clean_data_dir/'cleaned_location.csv'
order_payments = clean_data_dir/'cleaned_order_payments.csv'
orders_items = clean_data_dir/'cleaned_orders_items.csv'
order_reviews = clean_data_dir/'cleaned_order_reviews.csv'
orders = clean_data_dir/'cleaned_orders.csv'
products = clean_data_dir/'cleaned_products.csv'
sellers = clean_data_dir/'cleaned_sellers.csv'
```

```python
orders_df = pd.read_csv(orders)
orders_df = orders_df.drop('Unnamed: 0',axis=1)
orders_df.sample(5)

payments_df = pd.read_csv(order_payments)
payments_df = payments_df.drop('Unnamed: 0',axis=1)
payments_df.sample(5)

orders_items_df = pd.read_csv(orders_items)
orders_items_df = orders_items_df.drop('Unnamed: 0',axis=1)
orders_items_df.sample(5)
```

```python
orders_items_df.isnull().sum()
```

```python
## Creating a new columns of total amount of by doing sum of price and fright value
orders_items_df['total_amount'] = orders_items_df['price'] + orders_items_df['freight_value']
```

```python
orders_items_df.sample(5)
```

```python
orders_df.columns
```

```python
orders_df.info()
```

```python
# fixing the dtype
date_col = [
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]

for col in date_col:
    orders_df[col] = pd.to_datetime(orders_df[col])
```

```python
orders_df.info()
```

```python

# creating a new columns in orders_df delivery_days usig order_purchase_timestamp and order_delivered-customer_date
orders_df = orders_df.rename(
    columns={
        'order_purchase_timestamp' : 'order_purchase',
        'order_approved_at' : 'approved_at',
        'order_delivered_carrier_date' : 'delivered_carrier_date',
        'order_delivered_customer_date' : 'delivered_customer_date',
        'order_estimated_delivery_date' : 'estimated_delivery_date'
    }
)



orders_df['delivery_days'] = (orders_df['delivered_customer_date'] - orders_df['approved_at']).dt.days
# orders_df.sample(5)
```

```python
# adding a new columns delivery_delay

orders_df ['delivery_delay_days'] = (orders_df['delivered_customer_date'] - orders_df['estimated_delivery_date']).dt.days

# orders_df = orders_df.drop('delivery_delay',axis=1)

# adding a new columns is_delayed 
delivery_mask = orders_df['order_status'] == 'delivered'

orders_df['is_delayed'] = np.nan
# orders_df['is_delayed'] = (orders_df['estimated_delivery_date'] < orders_df['delivered_customer_date']).astype(int)

orders_df.loc[delivery_mask,'is_delayed'] = (
    orders_df.loc[
        delivery_mask, 'delivered_customer_date'
    ] > orders_df.loc[
        delivery_mask, 'estimated_delivery_date'
    ]
).astype(int)
```

```python
orders_df['order_status'].value_counts()
```

```python
orders_df.loc[
    orders_df['order_status'] != 'Invoiced','is_delayed'
].value_counts()

# orders_df.loc[orders_df['order_status'] != 'delivered', 'is_delayed'].value_counts()
```

```python
orders_df['is_delayed'].value_counts(dropna=False)
```

```python
orders_df.sample(5)
```

```python
customers_df = pd.read_csv(customers,index_col=0)
customers_df
```

```python
# merging customers and orders tables
orders_df = orders_df.merge(
    customers_df[['customer_id','customer_unique_id']],
    on='customer_id',
    how='left'
)

orders_df.duplicated().sum()
```

```python
orders_df['customer_order_count'] =orders_df.groupby('customer_unique_id')['order_id'].transform('count')
```

```python
orders_df['customer_order_count'].value_counts()

# adding bool columns bout is it a repeat customer
orders_df['is_repeat_customer'] = (orders_df['customer_order_count'] > 1).astype(int)
```

```python
orders_df.sample(5)
```

```python
# total spending per customer

orders_df.groupby('customer_unique_id')['order_id'].transform('count')
```

```python
feature_engineered_path = p_dir.FEATURE_ENGINEERED

orders_df.to_csv(feature_engineered_path/'featured_orders.csv',index=False)
```

#####  merging the products and category_name tables on category name col

```python


products_df = pd.read_csv(products,index_col=0)
products_df.sample(5)

category_name_df = pd.read_csv(category_name,index_col=0)
category_name_df = category_name_df.rename(columns={
    'product_category_name':'category_name',
    'product_category_name_english' : 'category_name_eng'
})
category_name_df.sample()
```

```python
products_df.duplicated().sum()
```

```python
products_df = products_df.merge(
    category_name_df[['category_name','category_name_eng']],
    on='category_name',
    how='left'
)

products_df.sample(5)
```

```python
products_df = products_df.drop(['length_cm','width_cm','height_cm'],axis=1)
products_df.sample(5)
```
